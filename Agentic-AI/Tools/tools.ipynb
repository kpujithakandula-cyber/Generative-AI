```python
from dotenv import load_dotenv
import os
load_dotenv()
model_name = 'google_genai:gemini-3.1-flash-lite'
```

```python
from langchain.agents import create_agent
```

```python
#creating tools
from langchain.tools import tool

@tool
def add(a: int|float, b: int|float) ->int|float:
    """ 
    Summary: 
        Adds two numbers and returns the result.

    Args:
        a (int|float): First number
        b (int|float): Second number
    Returns:
        int|float: Returns the sum of two numbers of a and b either integer or float
    """
    return a + b
```

```python
@tool
def subtract(a:int|float, b:int|float) -> int|float:
    """ 
    Summary:
        Subtracts one number from another number and returns the result.

    Args:
        a (int|float): First number
        b (int|float): Second number
    
    Returns:
        int|float: Returns the subtraction of two numbers of a and b either integer or float
    """
    return a - b
```

```python
@tool
def multiply(a: int|float, b: int|float) -> int|float:
    """ 
    Summary: Multiplication of two numbers.

    Args:
        a (int|float): First number
        b (int|float): Second number

    Retrurns:
        int|float: Product of two numbers of a and b either integer or float
    """
    return a * b
```

```python
agent = create_agent(
    model = model_name,
    tools = [add, subtract, multiply]
)
```

```python
from langchain.messages import AIMessage, HumanMessage, ToolMessage, SystemMessage
```

```python
def print_response(response):
    for message in response["messages"]:
        print(message)
```

```python
response = agent.invoke(
    {
        "messages": [
            HumanMessage("What is 20 + 10?"),
            HumanMessage("What is 45  - 40?"),
            HumanMessage("What is 20 * 30?")
        ]
    }
)
print_response(response)
```
