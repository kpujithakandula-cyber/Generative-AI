```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.1-flash-lite"
)
```

```python
messages = [
    SystemMessage("You are an expert in crime and investigations and you are well know detective."),
    HumanMessage("I'm aspiring to be a detective."),
    HumanMessage("What books should help in getting started?"),
    HumanMessage("Give me the output in tabular format"),
    HumanMessage("Suggest no more than 5 books")
]
```

```python
response = llm.invoke(messages)
```

```python
response.pretty_print()
```

```python
response.content[0]['text']
```

```python
for block in response.content_blocks:
    if block['type'] == 'text':
        print(block['text'])
```

```python
def print_response(response):
    for block in response.content_blocks:
        if block['type'] == 'text':
            print(block['text'])
```

## Zero Shot Prompting

```python
response = llm.invoke("What is intertia?")
```

```python
print(response)
```

```python
print_response(response)
```

## Few shots prompting

```python
messages = [
    HumanMessage("What is 7 - 8 + 6 ?"),
    AIMessage("""
    Step 1: 7 - 8 = -1
    Step 2: -1 + 6 = 5

    Answer: 5
    """),
    HumanMessage("What is 10 * 9 + 8 - 120 ?"),
    AIMessage("""
    Step 1: 10 * 9 = 90
    Step 2: 90 + 8 = 98
    Step 3: 98 - 120 = -22

    Answer: -22
    """),
    HumanMessage("What is 23 - 3 / 4 + 29?")
]
```

```python
response = llm.invoke(messages)
```

```python
print_response(response)
```

```python
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents import create_agent
cfg1 = {"configurable": {"thread_id": 1}}
cfg3 = {"configurable": {"thread_id": 3}}
agent = create_agent(
    model = "google_genai:gemini-3.1-flash-lite",
    checkpointer = InMemorySaver()
)
response = agent.invoke(
    {
    "messages": ["My name is Pujitha"]
    },
    cfg1
)
for message in response["messages"]:
    print(message)

```

```python
response = agent.invoke(
    {
    "messages": ["What is my name?"]
    },
    cfg1
)
for message in response["messages"]:
    print(message)
```

```python
response = agent.invoke(
    {
        "messages": ["Can you say what is my name?"]
    },
    cfg3
)
for message in response["messages"]:
    print(message)
```

```python
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()

store.put(("HR", "Hyderabad", "Chandu"), "favourites", {"food": "chicken biryani", "color": "red"})
store.put(("AI Engineer", "Hyderabad", "Pujitha"), "favourites", {"food": "laddu", "color": "black"})
```

```python
store.list_namespaces()
```

```python
store.get(('HR', 'Hyderabad', 'Chandu'), 'favourites')
```

```python
store.search(
    ('AI Engineer', 'Hyderabad', 'Pujitha'),
    query = 'laddu'
)
```

```python
agent = create_agent(
    model = 'google_genai:gemini-3.1-flash-lite',
    checkpointer = InMemorySaver(),   #short term memory
    store = store # long term memory
)
```
