```python
from langchain.tools import tool

@tool
def add(a, b):
    """
    Add two numeric values.

    Args:
        a: The first numeric value.
        b: The second numeric value.

    Returns:
        The sum of the two input values.
    """
    return a + b

@tool
def sub(a, b):
    """
    Subtract one numeric value from another.

    Args:
        a: The numeric value from which to subtract.
        b: The numeric value to subtract from the first value.

    Returns:
        The difference between the two input values.
    """
    return a - b

@tool
def mul(a, b):
    """
    Multiply two numeric values.

    Args:
        a: The first numeric value.
        b: The second numeric value.

    Returns:
        The product of the two input values.
    """
    return a * b

@tool
def div(a, b):
    """
    Divide one numeric value by another.

    Args:
        a: The dividend.
        b: The divisor.

    Returns:
        The quotient obtained by dividing the first value by the second value.

    Raises:
        ZeroDivisionError: If the divisor is zero.
    """
    return a / b

@tool
def mod(a, b):
    """
    Calculate the remainder after dividing one numeric value by another.

    Args:
        a: The dividend.
        b: The divisor.

    Returns:
        The remainder obtained from dividing the first value by the second value.

    Raises:
        ZeroDivisionError: If the divisor is zero.
    """
    return a % b
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = llm,
    tools = [add, sub, mul, div, mod]
)
```

```python
from langchain.messages import HumanMessage, ToolMessage, AIMessage, SystemMessage
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("What is 5 + 3?"),
]
```

```python
response = agent.invoke(
    {
        "messages": messages
    }
)
```

```python
response
```

```python
for message in response["messages"]:
    print(type(message))
    print(message.pretty_print())
```
